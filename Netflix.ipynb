```python
#1. Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('ggplot')

# Display all columns
pd.set_option('display.max_columns', None)
```

```python
#2. Load Dataset
df = pd.read_csv("netflix_titles.csv", encoding_errors='ignore')

print(df.shape)

df.head()
```

```python
#3. Dataset Information
df.info()

df.describe(include='all')
```

```python
#4. Remove Extra Columns
df = df.loc[:, ~df.columns.str.contains('^Unnamed')]

print(df.shape)
```

```python
print(df.columns)
```

```python
#5. Check Missing Values
df.isnull().sum()
```

```python
#6. Handle Missing Values

df['director'] = df['director'].fillna('Unknown')

df['cast'] = df['cast'].fillna('Unknown')

df['country'] = df['country'].fillna('Unknown')

df['rating'] = df['rating'].fillna(df['rating'].mode()[0])

df['date_added'] = df['date_added'].fillna('Not Available')
```

```python
#7. Remove Duplicate Records
print("Duplicates:", df.duplicated().sum())

df.drop_duplicates(inplace=True)

print(df.shape)
```

```python
df.info()
```

```python
#8. Convert Date Format
df['date_added'] = pd.to_datetime(
    df['date_added'],
    errors='coerce'
)
```

```python
#9. Feature Engineering
df['year_added'] = df['date_added'].dt.year

df['month_added'] = df['date_added'].dt.month_name()
```

```python
#10. Movies vs TV Shows
plt.figure(figsize=(6,4))

sns.countplot(
    x='type',
    data=df
)

plt.title("Movies vs TV Shows")
plt.show()
```

```python
df['type'].value_counts()
```

```python
#11. Top 10 Countries
countries = (
    df['country']
    .str.split(',')
    .explode()
    .str.strip()
)

top_countries = countries.value_counts().head(10)

print(top_countries)
```

```python
plt.figure(figsize=(10,5))

top_countries.plot(kind='bar')

plt.title("Top 10 Countries")
plt.xlabel("Country")
plt.ylabel("Count")

plt.show()
```

```python
#12. Top 10 Genres
genres = (
    df['listed_in']
    .str.split(',')
    .explode()
    .str.strip()
)

top_genres = genres.value_counts().head(10)

print(top_genres)
```

```python
plt.figure(figsize=(10,5))

top_genres.plot(kind='barh')

plt.title("Top Genres")

plt.show()
```

```python
#13. Content Added Over Years
yearly_content = (
    df['year_added']
    .value_counts()
    .sort_index()
)
```

```python
plt.figure(figsize=(12,5))

sns.lineplot(
    x=yearly_content.index,
    y=yearly_content.values
)

plt.title("Content Added Over Years")

plt.show()
```

```python
#14. Rating Distribution
plt.figure(figsize=(10,6))

sns.countplot(
    y='rating',
    data=df,
    order=df['rating'].value_counts().index
)

plt.title("Rating Distribution")

plt.show()
```

```python
#15. Top Directors
top_directors = (
    df[df['director'] != 'Unknown']
    ['director']
    .value_counts()
    .head(10)
)

print(top_directors)
```

```python
plt.figure(figsize=(10,5))

top_directors.plot(kind='barh')

plt.title("Top Directors")

plt.show()
```

```python
#16. Content Released Per Year
release_year = (
    df['release_year']
    .value_counts()
    .sort_index()
)
```

```python
plt.figure(figsize=(12,5))

plt.plot(
    release_year.index,
    release_year.values
)

plt.title("Release Year Trend")

plt.show()
```

```python
#17. Correlation Analysis
numeric_cols = df.select_dtypes(include=np.number)

corr = numeric_cols.corr()

plt.figure(figsize=(8,5))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Matrix")

plt.show()
```
